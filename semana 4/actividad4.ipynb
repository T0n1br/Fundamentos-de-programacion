```python
# Promediar cuantas calificaciones desea
def promedio(sumatoria, n):
    return sumatoria / n

n = int(input("Cuantas calificaciones quieres promediar"))


sumatoria = 0

print(f"Se van a pedir {n} calificaciones")
for c in range(1,n+1):
    calif = float(input(f"Dame la calificacion {c}"))
    sumatoria = sumatoria + calif

print(f"El promedio de {n} calificaciones es {promedio(sumatoria,n)}")
```

```python
filas = int(input("cuantas filas quieres"))
columnas = int(input("cuantas columnas quieres"))

matriz = []
for f in range(filas):
    fila = []
    for c in range(columnas):
        valor = 0
        if f == c:
            valor = 1
        elif f + c == filas - 1:
            valor = 1
        fila.append(valor)
    matriz.append(fila)

print("Impresion de la matriz:")
for f in range(filas):
    for c in range(columnas):
        print(matriz[f][c], end=" ")
    print()
  
```

```python
lista = []
contador = 1
while contador <= 10:
    valor = int(input("Dame  un numero entre -10 y 10: "))
    if valor >= -10 and valor <= 10:
        contador += 1
        lista.append(valor)
    else:
        print("El valor no es valido")

print(lista)
for i in range(len(lista)):
    if lista[i] < 0:
        for e in range(10-abs(lista[i])):
                    print(" ", end="")
        for j in range(abs(lista[i])):
            print("*", end="")    
        print("|")
    elif lista[i] > 0:
        for e in range(10):
            print(" ", end="")
        print("|", end="")
        for j in range(lista[i]):
            print("*", end="")
        print()
    else:
        print("|")
```
