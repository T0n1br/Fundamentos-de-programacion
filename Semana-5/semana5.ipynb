```python
candidatos = {
    "A": 0,
    "B": 0,
    "C": 0
}

cantidad = int(input("Cuantos votos van a ser?"))
conteo = 1

while conteo <= cantidad:
    voto = input(f"Ingresa el voto {conteo} (A, B, o C): ")
    if voto.upper() in candidatos:
        print(f"Voto valido para {voto.upper()}")
        candidatos[voto.upper()] += 1
        conteo += 1
    else:
        print("El voto no es valido")

    print("\nResultados")
for k,v in candidatos.items():
    print(f"Opcion {k} votos: {v}")
        
```

```python
sistema_activo = True

numeros = (5, 3, 2, 4, 1)

while sistema_activo == True:
    print("===== MENÚ =====")
    print("1. Tuplas")
    print("2. Diccionarios")
    print("3. Excepciones")
    print("4. Strings")
    print("5. Finalizar")

    opcion = (input("Por favor elige una opcion con su numero en entero: "))

    if opcion == "1":
        print("El numero que decidiste mostrar es: ", numeros[2])
        ad1 = int(input("Que numero deseas agregar: "))
        ad2 = int(input("Que numero 2 deseas agregar: "))
        numeros_nuevos = numeros + (ad1, ad2)
        print(numeros_nuevos)
        lista = list(numeros_nuevos)
        lista.sort()
        print(lista)
        def sumar_tupla(numeros):
            total = 0
            for numero in numeros:
                total = total + numero
            return total
        resultado = sumar_tupla(numeros_nuevos)
        print("La suma de la tupla es: ", resultado)
    elif opcion == "2":
        contactos = {
            "Juan": "4421234567",
            "Pedro": "4429876543",
            "Maria": "4425555555"
        }
        nombre_nuevo = input("Escribe el nombre del nuevo contacto: ")
        telefono_nuevo = input("Escribe el numero de telefono: ")
        contactos[nombre_nuevo] = telefono_nuevo
        print("Nombres de los contactos:")
        for nombre in contactos:
            print(nombre)
        def buscar_contacto(contactos, nombre):
            if nombre in contactos:
                return contactos[nombre]
        nombre_buscar = input("Que contacto deseas buscar: ")
        telefono = buscar_contacto(contactos, nombre_buscar)
        if telefono != None:
            print("El telefono es: ", telefono)
        else:
            print("El contacto no existe")
    elif opcion == "3":
        try:
            numero1 = int(input("Dame el primer numero entero: "))
            numero2 = int(input("Dame el segundo numero entero: "))
            suma = numero1 + numero2
            print("La suma es: ", suma)
            try:
                division = numero1 / numero2
                print("La division es: ", division)
            except ZeroDivisionError:
                print("No se puede dividir entre cero")
        except ValueError:
            print("Error: debes ingresar numeros enteros")
    elif opcion == "4":
        mensaje = "Hola mundo de la programacion"
        print("La longitud del mensaje es: ", len(mensaje))
        mensaje_mayusculas = mensaje.upper()
        print(mensaje_mayusculas)
        mensaje_nuevo = mensaje.replace("mundo", "universo")
        print(mensaje_nuevo)
        def cantidad_palabras(mensaje):
            palabras = mensaje.split()
            return len(palabras)
        cantidad = cantidad_palabras(mensaje)
        print("La cantidad de palabras es: ", cantidad)
    elif opcion == "5":
        sistema_activo = False
        print("Sistema finalizado")
    else:
        print("Opcion no valida")
```

```python
calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)
print("Tercera calificación: ", calificaciones[2])
cal1 = float(input("Nueva calificación 1: "))
cal2 = float(input("Nueva calificación 2: "))
nueva_tupla = calificaciones + (cal1, cal2)
print("Nueva tupla: ", nueva_tupla)
lista = list(nueva_tupla)
lista.sort(reverse=True)
print("Lista ordenada: ", lista)
def sumar_calificaciones(calificaciones):
    total = 0
    for calificacion in calificaciones:
        total = total + calificacion
    return total

suma = sumar_calificaciones(nueva_tupla)

print("Suma total: ", suma)
```
