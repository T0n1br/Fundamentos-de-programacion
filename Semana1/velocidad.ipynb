Algoritmo para calcular velocidad

```python
distancia = float(input("ingrese la distancia en km: "))
tiempo = float(input("ingrese el tiempo en horas: "))
velocidad = distancia / tiempo
print("La velocidad es de", velocidad, "km/h")
```
