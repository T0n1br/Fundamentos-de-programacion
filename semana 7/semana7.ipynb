```python
import time
import pdb

user = input("Cual es tu nombre:")
print("Bienvenido al Sistema de Citas Lila Vainilla,", user)

def pantalla_carga():
    print("Cargando sistema...")
    time.sleep(2)
    print("Sistema listo.")

pantalla_carga()

dia = int(input("Ingresa el dia: "))
mes = int(input("Ingresa el mes: "))
ano = int(input("Ingresa el año: "))
fecha = (dia, mes, ano)

precio_corte = 300
precio_maquillaje = 500
precio_spa = 1200

menu = [
    ["1", "Agendar cita"],
    ["2", "Ver citas"],
    ["3", "Agregar informacion"],
    ["4", "Ver archivos"],
    ["5", "Salir"]
]

sistema_activo = True

tiempo_inicio = time.time()

while sistema_activo == True:
    tiempo_actual = time.time()
    tiempo_inactivo = tiempo_actual - tiempo_inicio
    if tiempo_inactivo >= 600:
        print("\nHan pasado 10 minutos sin actividad.")
        continuar = input("Deseas continuar? Escribe si o no: ")
        if continuar == "si":
            tiempo_inicio = time.time()
        else:
            print("Regresando al inicio.")
            sistema_activo = False
            continue
    print("\n===== MENU PRINCIPAL =====")
    for opcion in menu:
        print(opcion[0], opcion[1])
    opcion = int(input("Que opcion quieres ")) 
    tiempo_inicio = time.time()
    if opcion == 1:
        print("\n===== AGENDAR CITA =====")
        nombre_cliente = input("Ingresa el nombre del cliente: ")

        servicios = [
            ["1", "Corte de cabello"],
            ["2", "Maquillaje"],
            ["3", "Spa"]
        ]

        for opcion_servicio in servicios:
            print(opcion_servicio[0], opcion_servicio[1])
        servicio = input("Selecciona el servicio: ")

        if servicio == "1":
            nombre_servicio = "Corte de cabello"
            precio = 300    
            anticipo = precio * 0.50
        elif servicio == "2":
            nombre_servicio = "Maquillaje"
            precio = 500
            anticipo = precio * 0.50
        elif servicio == "3":
            nombre_servicio = "Spa"
            precio = 1200
            anticipo = precio * 0.50
        else:
            print("Opcion no valida.")
            continue

        print("Nombre del cliente: ",nombre_cliente)
        print("El servicio es : ",nombre_servicio)
        print("El precio es de: ",precio)
        print("El anticipo es de: ",anticipo)

        pdb.set_trace

        with open("citas.txt", "a") as archivo:
            archivo.write("Fecha: " + str(fecha[0]) + "/" + str(fecha[1]) + "/" + str(fecha[2]) + "\n")
            archivo.write("Cliente: " + nombre_cliente + "\n")
            archivo.write("Servicio: " + nombre_servicio + "\n")
            archivo.write("Precio: " + str(precio) + "\n")
            archivo.write("Anticipo: " + str(anticipo) + "\n")
            archivo.write("-------------------------\n")

        with open("operaciones.txt", "a") as archivo:
            archivo.write("Fecha: " + str(fecha[0]) + "/" + str(fecha[1]) + "/" + str(fecha[2]) + "\n")
            archivo.write("Cita registrada\n")

        with open("clientes.txt", "a") as archivo:
            archivo.write(nombre_cliente + "\n")

    elif opcion == 2:
        try:
            with open("citas.txt", "r") as archivo:
                for linea in archivo:
                    print(linea.strip())

        except FileNotFoundError:
            print("No existe el archivo de citas.")

        with open("operaciones.txt", "a") as archivo:
            archivo.write("Fecha: " + str(fecha[0]) + "/" + str(fecha[1]) + "/" + str(fecha[2]) + "\n")
            archivo.write("Citas consultadas\n")

    elif opcion == 3:
        print("\n===== AGREGAR INFORMACION =====")
        print("1. Agregar cliente")
        print("2. Agregar operacion")
        print("3. Regresar")

        opcion_agregar = int(input("Que opcion quieres: "))

        if opcion_agregar == 1:
            nombre_cliente = input("Ingresa el nombre del cliente: ")
            with open("clientes.txt", "a") as archivo:
                archivo.write(nombre_cliente + "\n")
            print("Cliente guardado correctamente.")
        elif opcion_agregar == 2:
            operacion = input("Escribe la operacion realizada: ")
            with open("operaciones.txt", "a") as archivo:
                archivo.write(operacion + "\n")
            print("Operacion guardada correctamente.")
        elif opcion_agregar == 3:
            print("Regresando al menu principal.")
        else:
            print("Opcion no valida.")

    elif opcion == 4:
        print("\n===== ARCHIVOS DISPONIBLES =====")
        print("1. citas.txt")
        print("2. clientes.txt")
        print("3. servicios.txt")
        print("4. operaciones.txt")
        nombre_archivo = input("Escribe el nombre del archivo: ")
       
        try:
            with open(nombre_archivo, "r") as archivo:
                contenido = archivo.read()
            print("\n===== CONTENIDO DEL ARCHIVO =====")
            print(contenido)

        except FileNotFoundError:
            print("Error: el archivo no existe.")

        except PermissionError:
            print("Error: no tienes permiso para abrir este archivo.")
    elif opcion == 5:
        sistema_activo = False
        print("Gracias por utilizar el Sistema de Citas Lila Vainilla.")
    else:
        print("Opcion no valida")
```
