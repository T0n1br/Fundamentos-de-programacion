# Calculadora Digital ejercicio 1

Para este problema, primmero tuve que revisar los codigos base en el curso y documento que comppartio el profesor, dado que ya los entendia hice una practica imaginaria del programa con su solucion, procedi a realizarlo pidiendo primero el usuario por medio del input que es para solicitar que un usuario interactue y de caddena de texto, imprimir con print el saludo y la cadena del nombre de usuario, procedi a darle nmbre a las variables donde se va a guardar la informacion que el usuario comparta, imprimi un mensaje para que sepa que ya voy a calcular el tiempo, y le di valor a otra variable con la suma de todos sus tiempos en plataformas, utilize float por si eran tiempos inexactos, ya con el resultado imprimi un mensaje con su resultado, y con otra variable con operacion, para calcular el porcentaje de su tiempo total que representa en su dia, y imprimi tambien su resultado.

```python
username = input("Enter your username: ")
print("Hello,", username,"! Welcome to the program.")

ig = input("Cuanto tiempo le dedicas a instagram por dia? ")
tiktok= input("Cuanto tiempo le dedicas a tiktok por dia? ")
fb= input("Cuanto tiempo le dedicas a facebook por dia? ")
vg= input("Cuanto tiempo le dedicas a videojuegos por dia? ")
yt= input("Cuanto tiempo le dedicas a youtube por dia? ")

print("Listo ", username, ", ahora vamos a calcular cuanto tiempo le dedicas a las redes sociales y videojuegos por dia.")
tiempo_total = float(ig) + float(tiktok) + float(fb) + float(vg) + float(yt)
print("el tiempo que le dedicas al uso de plataformas es ", tiempo_total, "horas por dia.")

porcentaje = (tiempo_total / 24) * 100
print("Esto representa un ", porcentaje,"% de tu dia.")
```

 # Ejercicio 1 extra

 Para este problema declare las variables de la informacion que iba a solicitar, usando tambien float porque ya que estoy pidiendo cantidades pueden llevar decimal, y el input para que el usuario me de la informacion, ya teniendo esa informacion declare otras varables con las operaciones para calcular lo solicitado, y imprimir su propina y el total a pagar y entre las personas que sean.

```python
monto = float(input("Ingrese el monto de la cuenta: "))
porcentaje_propina = float(input("Ingrese el porcentaje de propina que desea dejar: "))
personas = int(input("Ingrese el numero de personas que van a pagar la cuenta: "))

propina = monto * porcentaje_propina / 100
total = monto + propina
total_por_persona = total / personas
print("La propina es de", propina, "y el total a pagar es de", total)
print("Cada persona debe pagar", total_por_persona)
```

# Ejercicio extra 2

Para este problema declare la variable con int ya que minutos si iban a ser enteros, declare las variables con las operaciones para conversion, y imprimi el resultado con la operacion ya realizada con la informacion que compartio.


```python
minutos = int(input("Ingrese el total en minutos: "))
dias = minutos // 1440
horas = (minutos % 1440) // 60
minutos_restantes = minutos % 60
print("El total es de", dias, "dia(s)", horas, "hora(s) y", minutos_restantes, "minutos restantes.")
```

 # Extra 3

 Para este problema declare mis variables solicitando la informacion de las notas obtenidas en los parciales con int ya que son enteros, y declare una variable usando float para un promedio exacto incluyendo decimales, finalice imprimiendo el promedio obtenido.
 

```python
parcial1 = int(input("Ingrese la nota del primer parcial: "))
parcial2 = int(input("Ingrese la nota del segundo parcial: "))
parcial3 = int(input("Ingrese la nota del tercer parcial: "))
promedio = float((parcial1 + parcial2 + parcial3) / 3)
print(f"El promedio de los tres parciales es: {promedio:.2f}")
```
