# Ejercicio 9

```python
monto=float(input("Cual es el monto de compra?"))
if monto<=100:
    print("tu total es de",monto)
elif 100<monto<=499:
    print("tu total es de",monto * .90)
else:
    print("tu total es de",monto * .80)

```
