# Avance de Proyecto-Fase 1

## Análisis organizacional:
La empresa que identifique es una empresa de belleza y cuidado personal bajo el nombre de Lila Vainilla, esta empresa se dedica a la venta de productos de maquillaje, cuidado capilar, cuidado de la piel, cremas, perfumes, etc. Sus necesidades es que tiene que hacer citas manualmente, al momento de agendar se pierde tiempo por atención al cliente, economía por inasistencias, empalmes en los horarios.

## Definición del problema:
El problema a resolver es automatizar su sistema de agendas, calculando de forma automática los costos y solicitar un anticipo del 50% para evitar cancelaciones asegurando al cliente.

## Reglas del negocio: 
1.	Costos fijos de cada servicio predeterminados y fijos.
2.	Evitar cancelaciones solicitando un anticipo del 50%.
3.	El sistema tiene que permitir agendar múltiples citas en una misma sesión hasta que el gerente o encargado en turno decida terminar el programa.

## Clasificación de datos:
1. Enteros(int): para las seleccion de las opciones en los servicios ej:(1.corte,2.permanente,3.tinte,etc.).
2. Cadenas (str): para almacenar el nombre del cliente y el servicio seleccionado.
3. Float: para calcular los precios con anticipo que puede incluir. decimales con los totales.
4. Booleanos: para los ciclos.

## Operadores de lenguaje:
1. Aritmeticos: para calcular el anticipo (*.50).
2. Relacionales: se utiliza (==) para las opciones del menu
3. Logicos: por si se necesita para validar entradas que esten dentro del rango (or) y (and).

## Estructuras de control: 
1. Iterativas: se utiliza para las citas que se deseen agendar o opciones que quieran agregar (while) y (for).
2. Condicionales: se utiliza para la toma de decisiones del usuario (if), (elif), y (else).
