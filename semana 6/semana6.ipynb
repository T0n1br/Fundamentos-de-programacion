```python
bateria = float(input("Bateria: "))

velocidad1 = float(input("Velocidad 1: "))
km1 = float(input("Km 1: "))

velocidad2 = float(input("Velocidad 2: "))
km2 = float(input("Km 2: "))

velocidad3 = float(input("Velocidad 3: "))
km3 = float(input("Km 3: "))

if velocidad1 == 60:
    consumo1 = 0.15
elif velocidad1 == 90:
    consumo1 = 0.22
else:
    consumo1 = 0.30

if velocidad2 == 60:
    consumo2 = 0.15
elif velocidad2 == 90:
    consumo2 = 0.22
else:
    consumo2 = 0.30

if velocidad3 == 60:
    consumo3 = 0.15
elif velocidad3 == 90:
    consumo3 = 0.22
else:
    consumo3 = 0.30

energia1 = km1 * consumo1
energia2 = km2 * consumo2
energia3 = km3 * consumo3
energia = energia1 + energia2 + energia3
porcentaje = energia / bateria * 100

restante = bateria - energia

km_restantes = restante / 0.15

print("Energia usada:", energia)
print("Porcentaje usado:", porcentaje)
print("Km restantes:", km_restantes)
```

ej 14

```python
def buscar_maximo(lista):
    mayor = lista[0]

    for numero in lista:
        if numero > mayor:
            mayor = numero

    return mayor


numeros = [7, 3, 9, 2]

resultado = buscar_maximo(numeros)

print("El elemento maximo es:", resultado)
```

ej 18

```python
def factorial(numero):

    if numero == 0:
        return 1
    else:
        return numero * factorial(numero - 1)


numero = int(input("Numero: "))

resultado = factorial(numero)

print("El factorial de", numero, "es:", resultado)
```
