# Cobro de entradas del museo

```python
visitantes = 1
costo_total = 0

while True:
    entrada = input(f"Visitante {visitantes}: ¿Cuál es tu edad? o escribe salir para terminar con tus visitantes")
    if entrada == ("salir"):
        break
    else:
        edad =float(entrada)
        visitantes = visitantes + 1
        if edad <= 3:
            print("Tu entrada es gratuita")
            costo = 0
        elif 3 < edad <= 17:
            costo = 30
            print("El costo es de 30")
        else:
            costo = 45
            print("El costo es de 45")

        visitante = input("Que tipo de visitante eres? adulto m, estudiante, profesor")

        if visitante == "adulto mayor":
            total = costo * .88
            costo_total = costo_total + total
            print("Tu descuento es del 12%, tu total es de", total)
        elif visitante == "estudiante" or visitante == "profesor":
            total = costo * .90
            costo_total = costo_total + total
            print("Tu descuento es del 10%, tu total es de", total)
        else:
            print("tu costo es general sin descuento")


print("fin del programa su total conjunto es de:",costo_total)

        
```

```python
visitantes = int(input("Cuantos visitantes son?"))

adultos = 0
suma_edades = 0
for(i) in range(1, visitantes + 1):
    edad = int(input(f"cual es la edad del visitante {i}? "))

    suma_edades = suma_edades + edad

    if edad >= 18:
        adultos = adultos + 1

prom = suma_edades / visitantes
print("Adultos:",float(adultos))
print("El promedio de las edades es",float(prom))

```
