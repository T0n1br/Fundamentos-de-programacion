# Ejercicio 15

```python
suma = 0
contador = 0

while True:
        cals = input("Dame tu calificacion (o escribe 'salir' para terminar): ")
        if cals == "salir":
            break
        cals = float(cals)
        suma = suma + cals
        contador = contador + 1

if contador > 0:
    promedio = suma / contador
    print(f"El promedio es: {promedio}")
```
